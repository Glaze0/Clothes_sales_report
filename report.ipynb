## Correcting Data Types 

```python
# Converting order_date and delivery_date to pandas date-time format

orders["order_date"] = pd.to_datetime(orders["order_date"], format="%Y-%m-%d")
orders['delivery_date'] = pd.to_datetime(orders['delivery_date'], format='%Y-%m-%d')

orders.info()
```

Finding -> What you should investigate
- 383 customers have no recorded orders                                 -> Inactive customers or incomplete order data?
- 7 orders have no corresponding sales lines                            -> Missing transaction records?
- 27 catalogue product IDs are not represented in sales                 -> Products never sold or missing transactions?
- All 5,000 sales prices differ from the corresponding catalogue prices	-> Discounts, price changes, or incorrect mapping?
- Order payment and sales-line totals differ substantially              -> Different units, definitions, or incorrect data?
- 7 order-product combinations appear more than once                    -> Repeated product lines or duplicate transactions?

# Advance EDA

```python
sales_merg = sales.merge(
    orders,
    on='order_id',
    how='left',
    validate='many_to_one'
)

sales_merg = sales_merg.merge(
    customers,
    on='customer_id',
    how='left',
    validate='many_to_one'
)

sales_merg = sales_merg.merge(
    products,
    on='product_id',
    how='left',
    validate='many_to_one'
)
```

```python
sales_merg.columns
```

```python
# Analysis 1: Which age group generates the highest sales?
sales_merg.groupby('age_group').agg(
    no_customers = ('customer_id','count'),
    total_order = ('order_id','count'),
    total_unit_p = ('quantity_x','sum'),
    total_rec_sales = ('total_price','sum'),
    avg_sale_per_customer = ('total_price','mean')
)
```

```python
sns.barplot(data = sales_merg, x = 'age_group', y = 'total_price', errorbar=None, estimator=sum)
plt.title('Total sales value across age groups')
```

```python
# Analysis 2: Which states generate the highest sales?
sales_merg.groupby('state').agg(
    unique_customer=('customer_id','nunique'),
    unique_order = ('order_id','nunique'),
    total_unit_sold = ('quantity_x', 'sum'),
    total_sales_value = ('total_price', 'sum')
)
```

```python
fig, ax = plt.subplots(1,2,figsize = [25,6])

sns.barplot(data = sales_merg, y = 'state', x = 'total_price', estimator=sum, ax=ax[0], errorbar=None)
ax[0].set_title('Total sales by state')

sns.barplot(data = sales_merg, y = 'state', x = 'order_id', estimator='nunique', ax=ax[1], errorbar=None)
ax[1].set_title('Number of Orders by state')
```

```python
# Analysis 3: Which products perform best?

sales_merg.groupby('product_name').agg(
    total_quan_sold = ('quantity_x','sum'),
    total_sales_value = ('total_price','sum'),
    no_sales_line = ('sales_id','count')
)
```

```python
sales_merg.groupby('product_type').agg(
    total_quan_sold = ('quantity_x','sum'),
    total_sales_value = ('total_price','sum'),
    no_sales_line = ('sales_id','count')
)
```

```python
sales_merg.groupby('size').agg(
    total_quan_sold = ('quantity_x','sum'),
    total_sales_value = ('total_price','sum'),
    no_sales_line = ('sales_id','count')
)
```

```python
sales_merg.groupby('colour').agg(
    total_quan_sold = ('quantity_x','sum'),
    total_sales_value = ('total_price','sum'),
    no_sales_line = ('sales_id','count')
)
```

```python
top10 = sales_merg.groupby('product_name').agg(
    total_sales_value = ('total_price','sum'),
).nlargest(10,'total_sales_value')

sns.barplot(data = top10, x = 'product_name', y = 'total_sales_value')
plt.xticks(rotation=75)
plt.title('Top 10 sold product')
```

```python
top10 = sales_merg.groupby('product_name').agg(
    t_quantity_sold = ('quantity_x','sum'),
).nlargest(10,'t_quantity_sold')

sns.barplot(data = top10, x = 'product_name', y = 't_quantity_sold')
plt.xticks(rotation=75)
plt.title('Top 10 sold product by Quantity')
```
